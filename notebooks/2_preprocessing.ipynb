```python
df_prep["Hour_sin"] = np.sin(2 * np.pi * df_prep["Hour"] / 24)

df_prep["Hour_cos"] = np.cos(2 * np.pi * df_prep["Hour"] / 24)
```

```python
X = df_prep.drop(columns=["Class"])
y = df_prep["Class"]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=42
)
```

```python
###mod
# Features that require a log transformation
amount_features = ["Amount"]

# Features that only require standard scaling
numeric_features = (
    ["Time"]
    + [f"V{i}" for i in range(1, 29)]
    + ["Hour_sin", "Hour_cos"]
)

print("Amount features:", amount_features)
print("Number of numeric features:", len(numeric_features))
print("Total features:", len(amount_features) + len(numeric_features))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer, StandardScaler
```

```python
amount_pipeline = Pipeline([
    ("log1p", FunctionTransformer(np.log1p, feature_names_out="one-to-one")),
    ("scaler", StandardScaler())
])
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("amount", amount_pipeline, amount_features),
        ("numeric", StandardScaler(), numeric_features)
    ],
    remainder="drop"
)
```

```python
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)
```

```python
# 2. Get feature names without transformer prefixes (e.g., 'amount__Amount' -> 'Amount')
feature_names = [
    col.split("__")[-1] for col in preprocessor.get_feature_names_out()
]
```

```python
# 3. Convert back to DataFrames, restoring original indices
X_train_df = pd.DataFrame(
    X_train_processed,
    columns=feature_names,
    index=X_train.index
)

X_test_df = pd.DataFrame(
    X_test_processed,
    columns=feature_names,
    index=X_test.index
)
```

```python
print(X_train_df.describe().T.loc[
    ["Time", "Amount", "Hour_sin", "Hour_cos"],
    ["mean", "std", "min", "max"]
])
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)

X_train_smote, y_train_smote = smote.fit_resample(
    X_train_df,
    y_train
)

print("X_train_smote:", X_train_smote.shape)
print("y_train_smote:", y_train_smote.shape)

print("\nSMOTE class distribution:")
print(y_train_smote.value_counts())
```

```python
X_train_smote.to_csv("../data/Processed/X_train_smote.csv", index=False)
```
